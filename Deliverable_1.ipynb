```python
import sys

import matplotlib.pyplot as plt
import requests

sys.path.append("..")
with open("../key_frost.py", "r", encoding="utf-8") as f:
    client_id = f.read().strip()

   

endpoint = 'https://frost.met.no/observations/v0.jsonld'
parameters = {
    'sources': 'SN17850',
    'elements': 'mean(air_temperature P1D)',
    'referencetime': '2025-01-01/2025-12-31',
}

frost_response = requests.get(endpoint, params=parameters, auth=(client_id, ""))

plt.plot
plt.show
print(frost_response)
```
